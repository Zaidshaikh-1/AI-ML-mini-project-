```python
# ---------- STEP 1: LOAD DATA (this is where your 2 files go) ----------
import pandas as pd, re

# The files are actually comma-separated (CSVs) with quotes around text. We parse them with sep=",".
train = pd.read_csv("drugsComTrain_raw.tsv", sep=",", on_bad_lines="skip")  # training file
test  = pd.read_csv("drugsComTest_raw.tsv", sep=",", on_bad_lines="skip")   # testing file

# we only need the review text and the rating
train = train[["review", "rating"]].dropna()
test  = test[["review", "rating"]].dropna()

# ---------- STEP 2: TURN RATINGS INTO LABELS ----------
def to_label(r):
    if r <= 4: return "negative"
    if r <= 6: return "neutral"
    return "positive"

train["label"] = train["rating"].apply(to_label)
test["label"]  = test["rating"].apply(to_label)
print(train["label"].value_counts())  # check class balance (put this in your report)

# ---------- STEP 3: CLEAN THE TEXT ----------
import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords

stop = set(stopwords.words("english")) - {"not", "no", "nor"}  # keep negations, they matter for sentiment

def clean(text):
    text = str(text).replace("&#039;", "'")         # fix HTML apostrophes in this dataset
    text = re.sub(r"<.*?>", " ", text)         # remove HTML tags
    text = re.sub(r"[^a-z\s]", " ", text.lower())  # lowercase, keep letters only
    return " ".join(w for w in text.split() if w not in stop)

train["clean"] = train["review"].apply(clean)
test["clean"]  = test["review"].apply(clean)

# ---------- STEP 4: TF-IDF (text -> numbers) ----------
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=50000, ngram_range=(1, 2), min_df=5)
X_train = tfidf.fit_transform(train["clean"])
X_test  = tfidf.transform(test["clean"])
y_train, y_test = train["label"], test["label"]

# ---------- STEP 5: TRAIN LOGISTIC REGRESSION ----------
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000, class_weight="balanced")
model.fit(X_train, y_train)

# ---------- STEP 6: EVALUATE ----------
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))
print(classification_report(y_test, pred))

ConfusionMatrixDisplay.from_predictions(y_test, pred, cmap="Blues")
title = "Confusion Matrix"
plt.title(title)
plt.show()

# ---------- STEP 7 (OPTIONAL): TRY YOUR OWN REVIEW ----------
def predict_review(text):
    return model.predict(tfidf.transform([clean(text)]))[0]

print(predict_review("This medicine gave me terrible headaches and did not help at all"))
print(predict_review("Worked great, my pain is gone and no side effects"))
```
